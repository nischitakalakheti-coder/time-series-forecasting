```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings ('ignore')

print ("All libraries imported successfully")
```

```python
Url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
df = pd.read_csv(Url)

df.columns = ['Month','Passengers']
df['month'] = pd.to_datetime(df['Month'])

df.set_index('Month',inplace=True)
print("Dataset loaded sucessfully")
print(f"Total rows: {len(df)}")
print(f"Date range: {df.index.min()} to {df.index.max()}")
```

```python
print("FIRST 12 ROWS (Year 1949):")
print(df.head(12))
print("\n LAST 6 ROWS:")
print( df.tail(6))
```

```python

print(" BASIC STATISTICS:")
print(df.describe())

print(f"\nMissing values: {df.isnull().sum().values[0]}")
print(f"Min passengers: {df['Passengers'].min()} thousand")
print(f"Max passengers: {df['Passengers'].max()} thousand")
print(f"Average passengers: {df['Passengers'].mean():.0f} thousand")
```

```python
plt.figure(figsize=(14, 5))
plt.plot(df.index, df['Passengers'], color='steelblue', linewidth=2)
plt.title('Monthly Airline Passengers (1949-1960)', fontsize=16)
plt.xlabel('Year')
plt.ylabel('Passengers (thousands)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('step1_timeseries.png')
plt.show()
print("Chart saved as step1_timeseries.png")
```
